```python
import json
import pandas as pd
```

```python
file_name = "CORD-19-research-challenge/metadata.csv"
df = pd.read_csv(file_name)
df.head()
```

```python
df.info()
```

```python
sample_size = 10000
random_seed = 42
json_path = 'metadata_sample.json'
df_sample = df.sample(n=sample_size, random_state=random_seed)
```

```python
df_sample.info()
```

```python
df_sample_filtered = df_sample.dropna(axis=1, how="all").dropna(axis=0, subset=['title', 'abstract', 'publish_time', 'authors'])

df_sample_filtered.info()
```

```python
df_sample_filtered.head()
```

```python
df_sample_filtered.to_json(json_path, orient='records', indent=2)
```
